```python
from pyserini.index import LuceneIndexReader
import json

index_reader = LuceneIndexReader("wiki_index")
```

```python
tasks = [1, 2, 3, 4, 5]
datasets = ["arguana", "kialo", "opinionqa"]
datasets_dict = {
    "arguana": "Arguana",
    "kialo": "Kialo",
    "opinionqa": "OpinionQA"
}
K = 5
lambds_chosen = {"arguana": 0.4, "kialo": 0.4, "opinionqa": 0.5}
```

```python
for task in tasks:
    print(f"\\subsection{{Task {task}}}")
    print(f"\\begin{{itemize}}")
    for dataset in datasets:
        if task!=4:
            filename = f"results/task{task}/{dataset}/samples_{K}.jsonl"
        else:
            filename = f"results/task{task}_{lambds_chosen[dataset]}/{dataset}/samples_{K}.jsonl"
        with open(filename, 'r') as f:
            first_line = f.readline()

            data = json.loads(first_line)
            query = data['query']
            docs = data['output_ids']

            print("\\item")
            print(f"\\textbf{{Query ({datasets_dict[dataset]})}}: \"{query}\"")
            print("\\begin{itemize}")
            for docid in docs:
                print(f"\\item {' '.join(eval(index_reader.doc(docid).get('raw'))['contents'].split()[:15])}...")
            print("\\end{itemize}")
    print(f"\\end{{itemize}}")
```

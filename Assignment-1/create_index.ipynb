```python
import os
import pandas as pd
import json
from tqdm import tqdm
```

```python
parquet_dir = 'wiki_dpr_data'
jsonl_file = 'wiki_jsonl/psgs_w100.jsonl'

os.makedirs(os.path.dirname(jsonl_file), exist_ok=True)

with open(jsonl_file, 'w', encoding='utf-8') as out_f:
    for filename in os.listdir(parquet_dir):
        if filename.endswith('.parquet'):
            filepath = os.path.join(parquet_dir, filename)
            df = pd.read_parquet(filepath)
            for _, row in tqdm(df.iterrows(), total=len(df), desc=f'Processing {filename}'):
                doc = {
                    'id': str(row['id']),
                    'contents': row['text']
                }
                out_f.write(json.dumps(doc) + '\n')
```

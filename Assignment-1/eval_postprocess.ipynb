```python
import os
import json
```

```python
task = 1
K = [5, 10, 20]
datasets = ["arguana", "kialo", "opinionqa"]
num_rows = {
    "arguana": 750,
	"kialo": 774,
	"opinionqa": 882
}
```

```python
for dataset in datasets:
	for k in K:
		print(f"Processing Task {task}, Dataset: {dataset}, K: {k}")
		file_path = f"eval_results/task{task}/{dataset}/eval_{k}.jsonl"
		if os.path.exists(file_path):
			with open(file_path, 'r') as f:
				lines = f.readlines()
				assert len(lines) == num_rows[dataset], f"Expected {num_rows[dataset]} lines, but got {len(lines)} lines in {file_path}"
				results = [json.loads(line) for line in lines]
				flaks = [res['flak'] for res in results]
				wct_1s = [res['wct_1'] for res in results]
				wct_5s = [res['wct_5'] for res in results]
				wct_9s = [res['wct_9'] for res in results]
				avg_flak = sum(flaks) / len(flaks)
				avg_wct_1 = sum(wct_1s) / len(wct_1s)
				avg_wct_5 = sum(wct_5s) / len(wct_5s)
				avg_wct_9 = sum(wct_9s) / len(wct_9s)
				final_results = {
					"Average FL@K": avg_flak,
					"Average WC@0.1": avg_wct_1,
					"Average WC@0.5": avg_wct_5,
					"Average WC@0.9": avg_wct_9
				}
				os.makedirs(f"final_results/task{task}/{dataset}", exist_ok=True)
				with open(f"final_results/task{task}/{dataset}/results_{k}.json", 'w') as out_f:
					json.dump(final_results, out_f, indent=4)
				print(final_results)
```

```python
task = 3
K = [5, 10, 20]
datasets = ["arguana", "kialo", "opinionqa"]
num_rows = {
    "arguana": 750,
	"kialo": 774,
	"opinionqa": 882
}
```

```python
for dataset in datasets:
	for k in K:
		print(f"Processing Task {task}, Dataset: {dataset}, K: {k}")
		file_path = f"eval_results/task{task}/{dataset}/eval_{k}.jsonl"
		file_path_task_1 = f"eval_results/task1/{dataset}/eval_{k}.jsonl"
		if os.path.exists(file_path):
			with open(file_path, 'r') as f, open(file_path_task_1, 'r') as f1:
				lines = f.readlines()
				lines_task_1 = f1.readlines()
				assert len(lines) == num_rows[dataset], f"Expected {num_rows[dataset]} lines, but got {len(lines)} lines in {file_path}"
				assert len(lines_task_1) == num_rows[dataset], f"Expected {num_rows[dataset]} lines, but got {len(lines_task_1)} lines in {file_path_task_1}"
				results = [json.loads(line) for line in lines]
				results_task_1 = [json.loads(line) for line in lines_task_1]
				flaks = [res['flak'] for res in results]
				nflaks = [res['flak']/res1['flak'] if res1['flak'] != 0 else 0 for res, res1 in zip(results, results_task_1)]
				wct_1s = [res['wct_1'] for res in results]
				wct_5s = [res['wct_5'] for res in results]
				wct_9s = [res['wct_9'] for res in results]
				avg_flak = sum(flaks) / len(flaks)
				avg_wct_1 = sum(wct_1s) / len(wct_1s)
				avg_wct_5 = sum(wct_5s) / len(wct_5s)
				avg_wct_9 = sum(wct_9s) / len(wct_9s)
				final_results = {
					"Average FL@K": avg_flak,
					"Average nFL@K": sum(nflaks) / len(nflaks),
					"Average WC@0.1": avg_wct_1,
					"Average WC@0.5": avg_wct_5,
					"Average WC@0.9": avg_wct_9
				}
				os.makedirs(f"final_results/task{task}/{dataset}", exist_ok=True)
				with open(f"final_results/task{task}/{dataset}/results_{k}.json", 'w') as out_f:
					json.dump(final_results, out_f, indent=4)
				print(final_results)
```
